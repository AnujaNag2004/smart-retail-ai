```python
from pathlib import Path
import pickle

import cv2
import numpy as np

faces_folder = Path("../data/raw/faces")
model_folder = Path("../app/models")

training_images = []
training_labels = []
label_to_customer = {}

for label, customer_folder in enumerate(sorted(faces_folder.iterdir())):
    if not customer_folder.is_dir():
        continue

    label_to_customer[label] = customer_folder.name

    for image_path in customer_folder.glob("*.png"):
        image = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)
        training_images.append(image)
        training_labels.append(label)

recognizer = cv2.face.LBPHFaceRecognizer_create()
recognizer.train(training_images, np.array(training_labels))

recognizer.write(str(model_folder / "lbph_face_recognizer.yml"))

with open(model_folder / "face_db.pkl", "wb") as file:
    pickle.dump(label_to_customer, file)

print("Face model trained.")
print("Customers:", label_to_customer)
```
